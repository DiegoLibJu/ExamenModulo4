## Procesamiento de texto

```python
import string
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

def text_process(mess):
  """
 A cada palabra se le realiza lo siguiente:
  1. Elimina la puntuación.
  2. Elimina las stopwords.
  3. Devuelve una lista de palabras limpias.
  """
  STOPWORDS = stopwords.words("english") + ["IJK", "IDK", "BTW", "4", "2", "u", "im", "dont", "doin", "ü", "ur", "ure", "ASAP"]

  #se revisa la puntuación
  nopunc = [char for char in mess if char not in string.punctuation]

  #aplicamos un JOIN
  nopunc = "".join(nopunc)

  #separamos por palabras
  return ' '.join([word for word in nopunc.split() if word.lower() not in STOPWORDS])
```

```python
df.head()
```

```python
df["Text"] = df.Motivation.apply(text_process)
df.head()
```

# Parte 3

Modelado Predictivo, Evaluación y Matriz de Confusión

## División del conjunto de datos en entrenamiento y prueba (train/test split)

```python
import seaborn as sns
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

x = df.Text
y = df.Label
print(x.shape)
print(y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=0)

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## Vectorización de características de texto

Ejemplos: CountVectorizer o TfidfTransformer

```python
from sklearn.feature_extraction.text import CountVectorizer

vect = CountVectorizer()
vect.fit(x_train)

x_train_dtm = vect.transform(x_train)
x_train_dtm = vect.fit_transform(x_train)

print(type(x_train_dtm), x_train_dtm.shape)

```

## Matriz de Confusión

```python
#transformación de datos de prueba a matriz

x_test_dtm = vect.transform(x_test)
print(type(x_test_dtm), x_test_dtm.shape)
```

```python
#matriz de recuento a una frecuencia de término

from sklearn.feature_extraction.text import TfidfTransformer

tfidf_transformer = TfidfTransformer()
tfidf_transformer.fit(x_train_dtm)
tfidf_transformer.transform(x_train_dtm)
```

## Entrenamiento del modelo de clasificación (ej. Multinomial Naive Bayes u otro algoritmo adecuado).

```python
#Naïve Bayes

from sklearn.naive_bayes import MultinomialNB

nb = MultinomialNB()
nb.fit(x_train_dtm, y_train)
```

```python
#MÉTRICA

from sklearn import metrics

y_pred_class = nb.predict(x_test_dtm)
print("Precisión del modelo bajo sus métricas:")
print(metrics.accuracy_score(y_test, y_pred_class))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

metrics.confusion_matrix(y_test, y_pred_class)
```

```python
mat_bayes = confusion_matrix(y_test, y_pred_class)
labels = "Physics", "Medicine", "Peace", "Literature", "Chemistry", "Economics"
sns.heatmap(mat_bayes, square=True, annot=True, fmt="d", cbar=False, cmap="Greens",
            xticklabels=labels, yticklabels=labels)


plt.title("Matriz de confusión — Clasificación de categoría Nobel")
plt.xlabel("Predicción")
plt.ylabel("Verdadero")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Desarrollo y Despliegue de la Aplicación Interactiva en Streamlit

```python
%%writefile NobelPrice.py

!pip install streamlit

import numpy as np
import streamlit as st
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB


st.write(''' # Predicción de categoría de Premio Nobel con entrenamiento basado en idioma Inglés ''')
st.image("AlfredNobel.jpg", caption="October 21, 1833 - December 10, 1896")

st.header('Texto')

def user_input_features():
  # Entrada
  texto = st.text_input("Introduce el texto a evaluar")

  user_input_data = {'Text': texto}

  features = pd.DataFrame(user_input_data, index=[0])

  return features

df = user_input_features()

nobel =  pd.read_csv('df_nobelprice.csv', encoding='latin-1')
X = nobel.Text
y = nobel.Label

vect = CountVectorizer()
X_dtm = vect.fit_transform(X)

nb = MultinomialNB()
nb.fit(X_dtm, y)

df_dtm = vect.transform(df['Text'])
prediction = nb.predict(df_dtm)

#{'physics':0, 'medicine':1, 'peace':2, 'literature':3, 'chemistry':4, 'economics':5}
#'Physics', 'Medicine', 'Peace', 'Literature', 'Chemistry', 'Economics'
st.subheader('Predicción')
if prediction == 0:
  st.write('Física')
elif prediction == 1:
  st.write('Medicina')
elif prediction == 2:
  st.write('Paz')
elif prediction == 3:
  st.write('Literature')
elif prediction == 4:
  st.write('Química')
elif prediction == 5:
  st.write('Economía')
else:
  st.write('Sin predicción')
```
